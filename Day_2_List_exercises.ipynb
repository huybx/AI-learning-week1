```python
sales_week1=[10,30,15,20,5,0,30,50]
sales_week2=[11,3,7,9,9,0,10]
sales=[]
CN = input("doanh số của ngày Chủ Nhật tuần 2?:")
sales_week2.insert(7,int(CN))

sales=sales_week1
sales.extend(sales_week2)
best_day = str(max(sales)*1.5)
worst_day= str(min(sales)*1.5)
print(f"best_day profit is {best_day} $")
print(f"worst_day profit is {worst_day} $")
print(f'total profit of the best day and the worst day is {best_day + worst_day} $')
```
