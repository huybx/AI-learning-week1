```python
def main():
  phone_number=input("nhập vào một số điện thoại di động với định dạng bất kỳ: ")
  phone_number = phone_number.strip()
  for ch in {"(",")","-","."}:
    phone_number=phone_number.replace(ch,"")
  phone_number=phone_number.replace("084","0")

  print(f"{phone_number}")
if __name__=="__main__":
  main()
```
