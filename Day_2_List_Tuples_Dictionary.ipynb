```python
print("list lesson")
Friends = ['Phương','Tuấn','Phụng','Hùng']
print(Friends) # lấy hết cả danh sách
print(Friends[0], Friends[2])
print(len(Friends))
print(Friends[1:2]) #lấy từ phần tử thứ nhất, đến trước phần tử thứ hai
print(Friends[1:3])
print(Friends[:]) # lấy hết cả danh sách
print(Friends[:2]) #lấy từ phần tử thứ 0, đến trước phần tử thứ 2
print(Friends[-1]) #lấy phần tử thứ 4 tính từ cuối danh sách
#---
Nums=[99,100,10,7,2,1000]
print(Friends )
Friends.sort()
print(Friends)
Friends.sort(reverse=True)
print(Friends)

Nums.append(11)
Nums.insert(0,99)
Nums[4]=88
Nums.sort(reverse=True)
Nums.extend(Friends)
a=Nums.pop(1)
print(a)
Nums.remove(Nums[1])
print(a)


```
