```python
# ==================== BƯỚC 1: EDA - Titanic Dataset ====================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset Titanic
df = sns.load_dataset('titanic')

print("Load dataset thành công!")
print("Kích thước dữ liệu:", df.shape)
print("Các cột:", df.columns.tolist())
# 1. Xem 5 dòng đầu tiên
print(df.head())

# 2. Thông tin tổng quát
print(df.info())

# 3. Thống kê số
print(df.describe())
# ==================== PHÂN TÍCH BIẾN MỤC TIÊU ====================

target = 'survived'

print("=== PHÂN BỐ CỦA BIẾN MỤC TIÊU ===")
print(df[target].value_counts())
print("\nTỷ lệ phần trăm:")
print(round(df[target].value_counts(normalize=True) * 100, 2))

# Vẽ biểu đồ
plt.figure(figsize=(8,5))
sns.countplot(data=df, x=target, palette='Set2')
plt.title('Phân bố hành khách sống sót hay không (0 = Chết, 1 = Sống)')
plt.ylabel('Số lượng')
plt.show()
# Đánh giá mức độ lệch
ratio = df[target].value_counts(normalize=True)
print(f"\nTỷ lệ lớp thiểu số (sống sót): {ratio[1]*100:.2f}%")

if ratio.min() > 0.4:
    print("→ Dữ liệu khá cân bằng")
elif ratio.min() > 0.2:
    print("→ Dữ liệu lệch nhẹ")
else:
    print("→ Dữ liệu bị lệch")
```
