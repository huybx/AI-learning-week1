```python
# 🐾 Dog Bus Tracker — Challenge Steps
from typing import Final
from datetime import datetime, timezone
import pytz
timezone = pytz.timezone('Asia/Ho_Chi_Minh')
MAX_SEATS: Final = 10 #sức chứa tối đa
current_seat=0 #số khách hiện tại
Dog_Bus = {} #danh sách khách - dùng dictionary

def initial_Dog_Bus():
  i=1
  while i<=MAX_SEATS:
    Dog_Bus.update({i:{}})
    i=i+1

def pickup(): #thêm một hành khách và cập nhật số khách hiện tại.
  global current_seat
  name= input("tên thú cưng là gì: ")
  breed = input ("thú cưng thuộc giống nào: ")
  pickup_time =  datetime.now(timezone).strftime("%H:%M:%S")
  dropoff_time = input("thú cưng được gửi mấy tiếng: ")
  for i in range(1,MAX_SEATS + 1):
    if Dog_Bus[i]== {}:
      Dog_Bus.update({i:{"name":name,"breed":breed,"pickup_time":pickup_time,"dropoff_time":f"trả sau {dropoff_time} giờ"}})
      current_seat = current_seat+1
      break

def dropoff_name(name:str): #thả khách theo tên và cập nhật số khách còn lại
  global current_seat
  count=0
  for i in range(1,MAX_SEATS + 1):
    info=Dog_Bus[i]
    if info and info["name"] == name:
      print(f"{info['name']} đã xuống xe")
      Dog_Bus.update({i:{}})
      current_seat = current_seat-1
      count=count+1
      print(f"số chỗ trống hiện tại là {MAX_SEATS-current_seat}")
  if count==0:
      print(f"khách tên {name} không có trên xe")

def dropoff_seat(seat:int): #thả khách theo số ghế và cho biết số khách còn lại
  global current_seat
  info=Dog_Bus[seat]
  if info != {}:
    print(f"{info['name']} đã xuống xe")
    Dog_Bus.update({seat:{}})
    current_seat = current_seat-1
    print(f"số chỗ trống hiện tại là {MAX_SEATS-current_seat}")
  else:
    print(f"ghế {seat} này không có khách")


def main():
 print("Chào mừng bạn đến với chương trình Dog BUS Tracker!")
 initial_Dog_Bus()
 num=int(input("số khách sắp lên xe là bao nhiêu: "))
 if num <= (MAX_SEATS-current_seat) :
   while num>0:
    print(f"đã có {current_seat} khách trên xe")
    pickup()
    num=num-1
 else:
   print("xe quá tải, không nhận khách")

 dropoff_seat(3)  #trả khách ghế 3
 dropoff_seat(1)  # trả khách ghế 1
 pickup()         #thêm 1 khách
 print(Dog_Bus)   #in danh sách khách
 dropoff_name("Ngân")  #trả khách theo tên
 dropoff_name("Linh")  #trả khách theo tên
 pickup()              #thêm 1 khách
 print(Dog_Bus)          #in danh sách khách
if __name__ == "__main__":
    main()
```
