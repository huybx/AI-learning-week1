```python
def ghi_thong_tin(file_name:str):
  my_file= open(file_name,"w", encoding="utf-8")
  while (True):
    Name = input("Họ tên học viên:")
    Age = input("Tuổi:")
    City = input("Thành phố:")
    l=Name + ','+Age+','+City+'\n'
    my_file.write(l)
    cont=input('Have you got anyone else: (Y/N)')
    if cont=='N':
      break
  my_file.close()

def doc_thong_tin(file_name):
  with open (file_name,"r", encoding="utf-8") as file:
    return(file.readlines())
  # close(file) # Removed redundant close(file) as 'with open' handles it.

file_name = str(input("Bạn muốn tạo file tên là gì"))
ghi_thong_tin(file_name)

file_name = str(input("Bạn muốn đọc dữ liệu từ file tên là gì"))
try:
  student =doc_thong_tin(file_name)
except FileNotFoundError as err:
  print(err)
else:
 print("Những cô sinh viên xinh đẹp:")
 for i in student:
  i=i.split(',')
  name=i[0]
  age = int(i[1].strip()) #strip() loại khoảng trắng trong chuỗi và chuyển chuỗi thành số nguyên
  city = (i[2].strip())
  print(f"{name} ở {city} năm nay {age} tuổi")
 #else:
 # print("Dữ liệu sinh viên không đúng định dạng.")
finally:
  print("kết thúc chương trình")

```
